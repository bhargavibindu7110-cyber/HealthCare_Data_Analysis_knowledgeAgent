```python
from pathlib import Path
import pandas as pd

# Create project folders
PROJECT_DIR = Path("/content/healthcare_data_analysis")
KNOWLEDGE_DIR = PROJECT_DIR / "knowledge"
DATA_DIR = PROJECT_DIR / "data"

KNOWLEDGE_DIR.mkdir(parents=True, exist_ok=True)
DATA_DIR.mkdir(parents=True, exist_ok=True)

# File 1: Healthcare knowledge index
(PROJECT_DIR / "knowledge" / "index.md").write_text(
"""---
type: index
title: Healthcare Knowledge Index
---

# Healthcare Knowledge Index

Available topics:
- [[healthcare_metrics]]
- [[data_analysis]]
""",
encoding="utf-8"
)

# File 2: Healthcare metrics
(KNOWLEDGE_DIR / "healthcare_metrics.md").write_text(
"""---
type: concept
title: Healthcare Metrics
tags:
  - blood-pressure
  - glucose
  - cholesterol
---

# Healthcare Metrics

## Blood Pressure
Blood pressure measures the force of blood against artery walls.
Systolic pressure is measured when the heart contracts.
Diastolic pressure is measured when the heart relaxes.
Blood pressure is measured in mmHg.

## Blood Glucose
Blood glucose is the amount of glucose in the blood.
Interpretation depends on the test method, fasting status,
and clinical context.

## Cholesterol
Cholesterol is a fatty substance found in the blood.
Different cholesterol measurements provide information about
cardiovascular health.

## Missing Values
Missing values are fields with no recorded information.
They should be examined before statistical analysis.
""",
encoding="utf-8"
)

# File 3: Healthcare data analysis
(KNOWLEDGE_DIR / "data_analysis.md").write_text(
"""---
type: concept
title: Healthcare Data Analysis
tags:
  - statistics
  - data-quality
---

# Healthcare Data Analysis

Healthcare data analysis examines data to identify patterns
and summarize information.

## Mean
The mean is the sum of valid numerical values divided by
the number of valid observations.

## Median
The median is the middle value when observations are sorted.

## Data Quality
Data quality checks include missing values, duplicate rows,
invalid measurements, and incorrect data types.

## Healthcare Safety
Statistical analysis alone cannot diagnose a disease.
Clinical interpretation requires appropriate medical context
and qualified healthcare professionals.
""",
encoding="utf-8"
)

# Create fictional sample healthcare data
healthcare_data = {
    "patient_id": [101, 102, 103, 104, 105, 106],
    "age": [25, 45, 35, 60, 50, 28],
    "systolic_bp": [118, 145, 125, 160, 135, 110],
    "diastolic_bp": [78, 92, 82, 100, 88, 72],
    "glucose": [90, 130, 105, 160, 115, 85],
    "cholesterol": [170, 230, 190, 250, 210, 165],
}

df = pd.DataFrame(healthcare_data)
CSV_PATH = DATA_DIR / "healthcare.csv"
df.to_csv(CSV_PATH, index=False)

print("Project files created successfully!")
print("Project folder:", PROJECT_DIR)
print("\nFiles created:")
for file_path in sorted(PROJECT_DIR.rglob("*")):
    if file_path.is_file():
        print("-", file_path.relative_to(PROJECT_DIR))

print("\nSample healthcare dataset:")
display(df)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

def load_knowledge():
    documents = []

    for file_path in KNOWLEDGE_DIR.glob("*.md"):
        if file_path.name == "index.md":
            continue

        content = file_path.read_text(encoding="utf-8")

        # Remove YAML metadata at the beginning
        if content.startswith("---"):
            parts = content.split("---", 2)
            if len(parts) == 3:
                content = parts[2].strip()

        documents.append({
            "source": file_path.name,
            "text": content
        })

    return documents


def retrieve_knowledge(question, documents, top_k=2):
    if not documents:
        return []

    texts = [doc["text"] for doc in documents]

    vectorizer = TfidfVectorizer(
        stop_words="english",
        ngram_range=(1, 2)
    )

    matrix = vectorizer.fit_transform(texts + [question])
    question_vector = matrix[-1]
    document_vectors = matrix[:-1]

    scores = cosine_similarity(
        question_vector, document_vectors
    ).flatten()

    ranked_indices = scores.argsort()[::-1]
    results = []

    for index in ranked_indices:
        if scores[index] <= 0:
            continue

        results.append({
            "source": documents[index]["source"],
            "text": documents[index]["text"],
            "score": float(scores[index])
        })

        if len(results) >= top_k:
            break

    return results


documents = load_knowledge()

print("RAG retrieval system is ready!")
print("Knowledge documents:", len(documents))

# Test retrieval before connecting the AI model
test_results = retrieve_knowledge(
    "What is blood pressure?",
    documents
)

for item in test_results:
    print("\nSource:", item["source"])
    print("Similarity:", round(item["score"], 3))
    print(item["text"][:250])
```

```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

MODEL_NAME = "Qwen/Qwen2.5-0.5B-Instruct"

print("Loading AI model...")

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    torch_dtype="auto",
    device_map="auto"
)

model.eval()

print("AI model loaded successfully!")
print("Device:", model.device)
```

```python
def generate_ai_answer(question, retrieved_docs):
    if not retrieved_docs:
        return (
            "I could not find relevant information in the "
            "healthcare knowledge base. Please try a more "
            "specific question."
        )

    context = "\n\n".join(
        f"Source: {doc['source']}\n{doc['text']}"
        for doc in retrieved_docs
    )

    messages = [
        {
            "role": "system",
            "content": (
                "You are a healthcare data analysis knowledge "
                "assistant. Explain concepts in simple English. "
                "Use the supplied context, do not invent facts, "
                "and state when information is unavailable. "
                "Never diagnose diseases or prescribe treatment."
            )
        },
        {
            "role": "user",
            "content": (
                f"Knowledge context:\n{context}\n\n"
                f"Question: {question}\n\n"
                "Answer using the context."
            )
        }
    ]

    prompt = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        prompt,
        return_tensors="pt"
    ).to(model.device)

    with torch.inference_mode():
        output = model.generate(
            **inputs,
            max_new_tokens=220,
            do_sample=False,
            pad_token_id=tokenizer.eos_token_id
        )

    new_tokens = output[0][inputs["input_ids"].shape[1]:]

    return tokenizer.decode(
        new_tokens,
        skip_special_tokens=True
    ).strip()


def analyze_healthcare_data(data):
    numeric_data = data.select_dtypes(include="number")

    summary = {
        "Total records": len(data),
        "Total columns": len(data.columns),
        "Missing values": int(data.isna().sum().sum()),
        "Duplicate rows": int(data.duplicated().sum()),
    }

    for column in [
        "age",
        "systolic_bp",
        "diastolic_bp",
        "glucose",
        "cholesterol"
    ]:
        if column in data.columns:
            summary[f"Average {column}"] = round(
                data[column].mean(), 2
            )

    return summary


def answer_question(question):
    question_lower = question.lower()

    # Analyze questions about the actual dataset
    data_keywords = [
        "dataset", "average", "mean", "how many",
        "missing", "duplicate", "patient", "records",
        "glucose", "cholesterol", "blood pressure",
        "systolic", "diastolic", "age"
    ]

    if any(word in question_lower for word in data_keywords):
        summary = analyze_healthcare_data(df)

        if "missing" in question_lower:
            return (
                f"The dataset contains "
                f"{summary['Missing values']} missing values."
            )

        if "duplicate" in question_lower:
            return (
                f"The dataset contains "
                f"{summary['Duplicate rows']} duplicate rows."
            )

        if "glucose" in question_lower:
            return (
                f"The average glucose value in this sample "
                f"is {summary['Average glucose']}. "
                "This is a statistical summary, not a diagnosis."
            )

        if "cholesterol" in question_lower:
            return (
                f"The average cholesterol value in this sample "
                f"is {summary['Average cholesterol']}. "
                "This is a statistical summary, not a diagnosis."
            )

        if "systolic" in question_lower:
            return (
                f"The average systolic blood pressure in this "
                f"sample is {summary['Average systolic_bp']} mmHg. "
                "This is a sample statistic, not a diagnosis."
            )

        if "diastolic" in question_lower:
            return (
                f"The average diastolic blood pressure in this "
                f"sample is {summary['Average diastolic_bp']} mmHg. "
                "This is a sample statistic, not a diagnosis."
            )

        if "blood pressure" in question_lower:
            return (
                f"Average systolic pressure: "
                f"{summary['Average systolic_bp']} mmHg.\n\n"
                f"Average diastolic pressure: "
                f"{summary['Average diastolic_bp']} mmHg.\n\n"
                "These are sample averages, not individual diagnoses."
            )

        if "age" in question_lower:
            return (
                f"The average age in the sample is "
                f"{summary['Average age']} years."
            )

        if "average" in question_lower or "mean" in question_lower:
            return "Dataset summary:\n" + "\n".join(
                f"- {key}: {value}"
                for key, value in summary.items()
            )

        if "how many" in question_lower or "records" in question_lower:
            return (
                f"The dataset contains "
                f"{summary['Total records']} records."
            )

        if "dataset" in question_lower or "patient" in question_lower:
            return "Dataset summary:\n" + "\n".join(
                f"- {key}: {value}"
                for key, value in summary.items()
            )

    # Otherwise, use RAG to answer knowledge questions
    retrieved_docs = retrieve_knowledge(
        question,
        documents,
        top_k=2
    )

    answer = generate_ai_answer(question, retrieved_docs)

    sources = "\n\nSources used:\n" + "\n".join(
        f"- {doc['source']}"
        for doc in retrieved_docs
    )

    return answer + sources


print("AI generation and data analysis are ready!")
print(analyze_healthcare_data(df))
```

```python
import gradio as gr
import matplotlib.pyplot as plt


def chatbot_response(question):
    if not question or not question.strip():
        return "Please enter a question.", ""

    try:
        # General knowledge questions should use RAG
        q = question.lower().strip()

        knowledge_questions = [
            "what is blood pressure",
            "explain blood pressure",
            "what is glucose",
            "what is cholesterol",
            "what is mean",
            "what is median",
            "what are missing values",
            "what is healthcare data analysis",
            "what is data quality"
        ]

        if any(q.startswith(item) for item in knowledge_questions):
            docs = retrieve_knowledge(question, documents, top_k=2)

            answer = generate_ai_answer(question, docs)

            sources = "\n".join(
                f"- {doc['source']}" for doc in docs
            )

            return answer, sources or "No sources found."

        # Other questions go through the combined agent
        answer = answer_question(question)

        return answer, "See the answer for its data or knowledge context."

    except Exception as error:
        return f"An error occurred: {error}", ""


def make_chart():
    columns = [
        "age",
        "systolic_bp",
        "diastolic_bp",
        "glucose",
        "cholesterol"
    ]

    averages = df[columns].mean()

    fig, ax = plt.subplots(figsize=(9, 4))
    averages.plot(kind="bar", ax=ax)

    ax.set_title("Healthcare Dataset: Average Values")
    ax.set_ylabel("Average value")
    ax.set_xlabel("Measurement")
    ax.tick_params(axis="x", rotation=30)

    fig.tight_layout()
    return fig


with gr.Blocks(title="Healthcare Data Analysis Agent") as app:

    gr.Markdown(
        """
        # 🏥 Healthcare Data Analysis Knowledge Agent

        Ask healthcare knowledge questions, retrieve information
        from Markdown documents, and analyze a sample dataset.

        **Demo only:** The patient records are fictional.
        This application is not a diagnostic tool.
        """
    )

    with gr.Tab("Dashboard"):
        gr.Markdown("## Healthcare Statistics")

        summary = analyze_healthcare_data(df)

        gr.JSON(value=summary, label="Dataset Summary")

        gr.Markdown("## Sample Healthcare Dataset")
        gr.Dataframe(value=df, interactive=False)

        gr.Markdown("## Average Values Chart")
        gr.Plot(value=make_chart())

    with gr.Tab("AI Knowledge Agent"):
        question_box = gr.Textbox(
            label="Ask a question",
            placeholder=(
                "Example: What is blood pressure?"
            ),
            lines=2
        )

        ask_button = gr.Button("Ask AI")

        answer_box = gr.Textbox(
            label="Answer",
            lines=8
        )

        sources_box = gr.Textbox(
            label="Retrieved Knowledge Sources",
            lines=3
        )

        ask_button.click(
            fn=chatbot_response,
            inputs=question_box,
            outputs=[answer_box, sources_box]
        )

        question_box.submit(
            fn=chatbot_response,
            inputs=question_box,
            outputs=[answer_box, sources_box]
        )


app.launch(share=True)
```
